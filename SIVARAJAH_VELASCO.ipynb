Ce graphique qui représente la magnitude absolue des étoiles en fonction de leur indice de couleur (B-V), permet notamment de distinguer différentes populations stellaires telles que la séquence principale, les géantes, les supergéantes ou les naines blanches.

### 1.2 - Nature du problème

Nous disposons d’un ensemble de données non labellisées, c’est-à-dire sans étiquettes indiquant à quels types appartiennent les étoiles.

L’objectif est d’explorer comment des techniques de réduction de dimension (PCA) et de clustering non supervisé peuvent être appliquées pour segmenter automatiquement les étoiles en groupes cohérents, en s’appuyant sur leur position dans le diagramme HR.

### 1.3 - Sélection des modèles 

- **K-Means** : \
K-Means regroupe les données en K clusters en minimisant la distance entre les points et leur centre (centroïde) de cluster. L’algorithme part de centres initiaux choisis aléatoirement, puis les ajuste itérativement jusqu’à convergence.  Les clusters doivent avoir une forme sphérique et il est nécessaire de connaitre à l'avance le nombre de clusters.

- **DBSCAN** (Density-based Spatial Clustering of Applications with Noise) : \
DBSCAN regroupe les points en clusters denses séparés par des zones de faible densité. Il ne nécessite pas de connaître le nombre de clusters à l’avance. 

- **Gaussian mixture models** (GMM) : \
GMM suppose que les données sont générées par un mélange de distributions normales.


### 1.4 - Choix de la métrique pour évaluer les modèles

Comme il s'agit d'un apprentissage non supervisé, nous pouvons seulement utiliser le silhouette score qui permet de mesurer la qualité d’un clustering, c’est-à-dire à quel point les étoiles sont bien groupés dans leurs clusters respectifs.

### 1.5 - Justification de la labellisation des données et apprentissage supervisé

Les données à notre disposition sont non labellisées, ce qui limite considérablement les méthodes d'évaluation pour les modèles de clustering non supervisé. En l'absence de véritables clusters de référence, notre principal indicateur de performance reste le silhouette score. Pour cela, nous avons tenté de labelliser les données afin d'effectuer un apprentissage supervisé. On considérera le diagramme obtenu suite à un apprentissage supervisé comme notre diagramme de référence. ### 2.2 - Calcul des grandeurs physiques

Le diagramme HR représente la magnitude absolue des étoiles en fonction de leurs indices de couleur (B-V). On commence par déterminer la distance entre le Soleil et chaque étoile, en calculant la distance individuelle pour chacune d'elles.

$$ d = \frac{1}{p}$$
Où p correspond à la parallaxe ("Plx" dans le dataset).


On peut alors déterminer la magnitude absolue qui correspond à la magnitude apparente d'un objet céleste s'il se trouve à une distance de 10 parsecs (pc) de l'observateur :

$$ M = m - 5log_{10} \frac{d}{10pc}$$
Où m est la magnitude apparente de l'étoile ("Vmag" dans le dataset).



```python
## On calcule les 3 grandeurs dans data_pca_calcul afin de les utiliser pour la pca

# Copie pour éviter de modifier directement le dataset initial
data_calcul = data_true.copy()

# Filtrer les étoiles avec une parallaxe positive 
data_calcul = data_calcul[data_calcul['Plx'] > 0]

# Calcul de la distance (en pc) 
data_calcul['Distance'] = 1/data_calcul['Plx'] * 1000   # dans les données Plx est donné en mas donc il faut convertir en arcsec

# Calcul de la magnitude absolue 
data_calcul['AbsoluteMag'] = data_calcul['Vmag'] - 5 * np.log10(data_calcul['Distance'] / 10)   

# Fonction pour calculer la température effective des étoiles
def bv_to_temp(bv):
    return 4600 * ((1 / (0.92 * bv + 1.7)) + (1 / (0.92 * bv + 0.62)))

# Ajouter la température aux données
data_calcul['Temperature'] = bv_to_temp(data_calcul['B-V'])

# Vérifier que les colonnes calculées sont bien dans data_pca_calcul
print("Distance est une colonne dans data_pca_calcul:", "Distance" in data_calcul)
print("AbsoluteMag est une colonne dans data_pca_calcul:", "AbsoluteMag" in data_calcul)
print("Temperature est une colonne dans data_pca_calcul:", "Temperature" in data_calcul)
data_calcul
```

## 3) Labellisation des données

Étant donné que les données d'étoiles ne sont pas préalablement labellisées, nous allons attribuer des catégories aux étoiles en nous basant sur notre connaissance de la physique des diagrammes HR. Sur ce diagramme, chaque catégorie d'étoile se situe dans une zone spécifique en fonction de son indice de couleur (B-V) et de sa magnitude absolue (AbsoluteMag). Nous allons ainsi tracer les courbes correspondant à chacune des catégories d'étoiles (Séquence principale, Géantes, Supergéantes, Naines blanches), et, en posant des conditions sur les indices de couleur, nous pourrons associer à chaque étoile un type basé sur sa position dans le diagramme.

```python
def mag_seq_princ(bv):
    return 4.95 * bv**3 - 12.07 * bv**2 + 13.91 * bv

def mag_geantes(bv):
    return -2.1 * bv**2.2 + 2.9 * bv + 0.5

def mag_supergiantes(bv):
    return -6 * np.exp(-bv) - 1.2 * bv**2 - 5

def mag_naines_blanches(bv):
    return -2.1 * (bv + 0.5)**2 + 5.9 * (bv + 0.5) + 9

bv_seq_princ = np.linspace(-0.2, 1.7, 100)
bv_geantes = np.linspace(0.8, 1.7, 100) 
bv_superg = np.linspace(-0.2, 1.7, 100) 
bv_naine_blanche = np.linspace(-0.2, 1.0, 100)

abs_mag_seq_princ = mag_seq_princ(bv_seq_princ)
abs_mag_geantes = mag_geantes(bv_geantes)
abs_mag_superg = mag_supergiantes(bv_superg)
abs_mag_naine_blanche = mag_naines_blanches(bv_naine_blanche)

# Tracer les courbes
plt.style.use('dark_background')
plt.figure(figsize=(10, 8))
plt.scatter(data_clean['B-V'], data_clean['AbsoluteMag'], c=data_clean['Temperature'], s=1, cmap="plasma", label='Étoiles')
plt.plot(bv_seq_princ, abs_mag_seq_princ, color='yellow', linestyle='--', linewidth=2, label='Séquence Principale')
plt.plot(bv_geantes, abs_mag_geantes, color='red', linestyle='--', linewidth=2, label='Géantes')
plt.plot(bv_superg, abs_mag_superg, color='orange', linestyle='--', linewidth=2, label='Supergéantes')
plt.plot(bv_naine_blanche, abs_mag_naine_blanche, color='cyan', linestyle='--', linewidth=2, label='Naines Blanches')
plt.gca().invert_yaxis()
plt.xlim(-0.5, 2.5)
plt.xlabel("B-V")
plt.ylabel("Magnitude Absolue")
plt.title("Courbes types sur le diagramme HR")
plt.legend()
plt.show()
```

Nous définissons une fonction **labelliser_etoile** qui, pour chaque étoile, utilise ses valeurs de B-V et de magnitude absolue pour déterminer à quel type d'étoile elle appartient. Cette fonction évalue la distance de chaque étoile par rapport aux courbes théoriques pour les différentes catégories d'étoiles. En fonction de la distance la plus faible, l'étoile se voit attribuer le type correspondant.

```python
def labelliser_etoile(row):
    bv = row['B-V']
    mag = row['AbsoluteMag']
    
    if np.isnan(bv) or np.isnan(mag):
        return np.nan

    distances = {}

    if -0.2 <= bv <= 1.7:
        distances['Séquence principale'] = abs(mag - mag_seq_princ(bv))
    else:
        distances['Séquence principale'] = np.inf

    if 0.8 <= bv <= 2.0:
        distances['Géante'] = abs(mag - mag_geantes(bv))
    else:
        distances['Géante'] = np.inf

    if -0.2 <= bv <= 1.7:
        distances['Supergéante'] = abs(mag - mag_supergiantes(bv))
    else:
        distances['Supergéante'] = np.inf

    if -0.2 <= bv <= 1.0:
        distances['Naine blanche'] = abs(mag - mag_naines_blanches(bv))
    else:
        distances['Naine blanche'] = np.inf

    return min(distances, key=distances.get)



data_clean.loc[:, 'Type_Étoile'] = data_clean.apply(labelliser_etoile, axis=1)



plt.figure(figsize=(10, 8))
sns.scatterplot(data=data_clean, x='B-V', y='AbsoluteMag', hue='Type_Étoile', palette='plasma', s=10)
plt.style.use('dark_background')
plt.gca().invert_yaxis()
plt.xlim(-0.5, 2.5)
plt.title("Diagramme HR - Visualisation des données labellisées")
plt.legend()
plt.show()

data_clean
```

### 6.3 - Optimisation du nombre de clusters

```python
## Optimisation du nombre de clusters sur les données d'entrainement
## Source [2] - voir bibliographie

inertias = []
cluster_range = range(1, 20)

for k in cluster_range:
    kmeans = KMeans(n_clusters=k, n_init=10, random_state=0)  # n_init : nombre d'initialisations (points de départ différents pour chaque initialisation)
    kmeans.fit(X_train_t_df)  
    inertias.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(cluster_range, inertias, marker='o', color="blue")
plt.xlabel("Nombre de clusters")
plt.ylabel("Inertie intra-cluster")
plt.title("Méthode du coude pour choisir le nombre de clusters (K-means)")
plt.grid(linestyle='--', color='gray', alpha=0.5)
plt.show()
```

On observe qu’au début, l’inertie diminue fortement lorsque le nombre de clusters augmente (jusqu’à environ 4). Ensuite, la diminution devient plus progressive. Ce changement de rythme correspond au "coude". On peut donc choisir le nombre optimal de clusters au niveau de ce coude, là où l’ajout de nouveaux clusters n’apporte plus d’amélioration significative. 

```python
plt.figure(figsize=(8, 5))
plt.plot(cluster_range, inertias, marker='o', color="blue")
plt.axvline(x=4, linestyle='--', color='pink')
plt.xlabel("Nombre de clusters")
plt.ylabel("Inertie intra-cluster")
plt.title("Méthode du coude pour choisir le nombre de clusters (K-means)")
plt.grid(linestyle='--', color='gray', alpha=0.5)
plt.show()
```

On peut généraliser à l'aide d'une fonction afin de modifier facilement les nombres de composantes et clusters.

```python
def pca_kmeans(nb_comp, nb_cluster):

    # Appliquer la PCA sur les ensembles de données
    pca_train = decomposition.PCA(n_components=nb_comp)   
    X_train_t = pca_train.fit_transform(X_train_scaled_df)
    X_train_t_df = pd.DataFrame(X_train_t, columns=[f'Col{i+1}' for i in range(nb_comp)])

    pca_test = decomposition.PCA(n_components=nb_comp) 
    X_test_t = pca_test.fit_transform(X_test_scaled_df)  
    X_test_t_df = pd.DataFrame(X_test_t, columns=[f'Col{i+1}' for i in range(nb_comp)])

    # Entraînement du modèle (données d'entrainement)
    kmeans = KMeans(n_clusters=nb_cluster, random_state=0, n_init=10)
    kmeans.fit(X_train_t_df)

    # Prédire les clusters (données d'entrainement)
    train_clusters_kmeans_val = kmeans.predict(X_train_t_df)

    # Faire une copie des données utilisées pour la PCA 
    X_train_hr = X_train.copy()
    X_train_hr = X_train_hr.reset_index(drop=True)

    # Ajouter les labels des clusters
    X_train_hr['Cluster'] = train_clusters_kmeans_val

    # Définir la figure 
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    fig.suptitle(f"Analyse PCA et K-Means (Entrainement et Test)\n{nb_comp} composantes, {nb_cluster} clusters", fontsize=16)

    # Diagramme HR (Entrainement) 
    sns.scatterplot(x=X_train_hr['B-V'], y=X_train_hr['AbsoluteMag'], hue=train_clusters_kmeans_val, palette='plasma', marker='.', s=10, ax=axes[0, 0])
    axes[0, 0].set_title(f"Diagramme HR - Entrainement")
    axes[0, 0].set_xlabel('B-V')
    axes[0, 0].set_ylabel('Magnitude absolue')
    axes[0, 0].invert_yaxis()
    axes[0, 0].set_xlim(-0.5, 2.5)

    # Graphe des clusters dans l'espace réduit par PCA (Entrainement) 
    sns.scatterplot(x=X_train_t_df['Col1'], y=X_train_t_df['Col2'], hue=kmeans.labels_, palette='plasma', s=10, ax=axes[0, 1])
    axes[0, 1].set_title(f"Clusters dans l'espace réduit PCA - Entrainement")
    axes[0, 1].set_xlabel("Composante principale 1")
    axes[0, 1].set_ylabel("Composante principale 2")

    # Silhouette score pour l'entrainement
    score_val = silhouette_score(X_train_t_df, train_clusters_kmeans_val)
    axes[0, 1].text(0.5, 0.9, f"Silhouette Score (Entrainement): {score_val:.3f}", horizontalalignment='center', transform=axes[0, 1].transAxes, fontsize=12, fontweight='bold')

    # Prédire les clusters (données de test)
    train_clusters_kmeans_test = kmeans.predict(X_test_t_df)

    # Faire une copie des données de test utilisées pour la PCA 
    X_test_hr = X_test.copy()
    X_test_hr = X_test_hr.reset_index(drop=True)

    # Ajouter les labels des clusters
    X_test_hr['Cluster'] = train_clusters_kmeans_test

    # Diagramme HR (Test) - Troisième sous-graphe
    sns.scatterplot(x=X_test_hr['B-V'], y=X_test_hr['AbsoluteMag'], hue=train_clusters_kmeans_test, palette='plasma', marker='.', s=10, ax=axes[1, 0])
    axes[1, 0].set_title(f"Diagramme HR - Test")
    axes[1, 0].set_xlabel('B-V')
    axes[1, 0].set_ylabel('Magnitude absolue')
    axes[1, 0].invert_yaxis()
    axes[1, 0].set_xlim(-0.5, 2.5)

    # Graphe des clusters dans l'espace réduit par PCA (Test) - Quatrième sous-graphe
    sns.scatterplot(x=X_test_t_df['Col1'], y=X_test_t_df['Col2'], hue=X_test_hr['Cluster'], palette='plasma', s=10, ax=axes[1, 1])
    axes[1, 1].set_title(f"Clusters dans l'espace réduit PCA - Test")
    axes[1, 1].set_xlabel("Composante principale 1")
    axes[1, 1].set_ylabel("Composante principale 2")

    # Silhouette score pour les données de test
    score_test = silhouette_score(X_test_t_df, train_clusters_kmeans_test)
    axes[1, 1].text(0.5, 0.9, f"Silhouette Score (Test): {score_test:.3f}", horizontalalignment='center', transform=axes[1, 1].transAxes, fontsize=12, fontweight='bold')

    plt.tight_layout()
    plt.subplots_adjust(top=0.9)  
    plt.show()
```

## 10) Analyse des résultats 

## 10.1 - K-Means

Dans un premier temps, nous avons appliqué l’algorithme de K-Means sur les données sans réduction de dimension par PCA. Les résultats obtenus montrent des scores de silhouette très faibles, avec une valeur de 0.0686 pour l’ensemble d’entraînement et 0.0691 pour l’ensemble de test. Ces scores indiquent une mauvaise qualité de séparation des clusters, suggérant que les données, dans leur forme initiale, ne sont pas bien adaptées à une segmentation claire par K-Means.

Pour améliorer les performances du clustering, nous avons exploré l’utilisation de la PCA (Analyse en Composantes Principales). L’objectif ici est de réduire la dimensionnalité des données tout en conservant l’essentiel de leur variance.

Nous avons ainsi implémenté une fonction permettant de tester K-Means pour différents nombres de composantes principales. Grâce à cette approche, nous avons observé que le silhouette score augmente significativement (jusqu'à atteindre environ 0.35) lorsque les données sont projetées sur deux composantes principales.

Nous avons également visualisé les diagrammes HR pour les ensembles d'entraînement et de test, en comparant les résultats obtenus avec et sans réduction de dimension par PCA. Dans les deux cas, nous constatons que les scores de silhouette sont très proches entre les jeux de données d'entraînement et de test. Cela indique une bonne stabilité du modèle et suggère que la structure du clustering détectée est  généralisable, même après séparation des données.

## 10.2 - DBSCAN

Le clustering avec l’ensemble des données sépare les étoiles en 4 groupes, ce qui est attendu, et donne un bon score numérique pour le silhouette score ($\approx 0.667$). Cependant, cette séparation ne correspond pas à une segmentation astrophysiquement significative des étoiles.

Lorsque l’on applique une séparation entre jeu d'entraînement et jeu de test, l’algorithme assigne presque tous les points au cluster -1, c’est-à-dire qu’ils sont considérés comme du bruit. Cela rend impossible le calcul du score de silhouette, car il faut au moins deux clusters définis pour l’évaluer. Comme les densités locales peuvent varier, les structures détectées dans l’ensemble d'entraînement ne se retrouvent plus dans l’ensemble de test.

Bien que le clustering après avoir fait une réduction de dimensionnalité avec PCA ait potentiellement simplifié l’espace des caractéristiques, elle n’a pas suffi à faire émerger des structures de densité exploitables par DBSCAN.

Le manque de structure bien définie entre les groupes denses et la sensibilité à la répartition des données, rend DBSCAN inadapté à ce type de problématique de classification stellaire, où les groupes ne sont pas clairement séparés par des régions de faible densité.

## 10.3 - GMM

Sans appliquer la PCA, les silhouette scores obtenus sont très faibles, ce qui indique que les clusters sont peu cohérents et mal séparés. Visuellement, les clusters apparaissent fortement mélangés sur les diagrammes HR. Cela confirme bien la mauvaise structuration des groupes détectés par ce modèle.

Cependant, après avoir appliqué la PCA (en réduisant à deux composantes principales), on observe un phénomène contradictoire : le silhouette score diminue encore, jusqu’à devenir négatif, ce qui suggère un mauvais clustering.

Pourtant, la visualisation des clusters semble bien plus convaincante : les groupes sont visiblement plus distincts et moins entrelacés suite à la PCA.

On peut expliquer cette incohérence entre le score et la visualisation du diagramme HR par le fait que GMM supposent des distributions gaussiennes en entrée, or la PCA a dûe les rendre moins gaussiennes dans l’espace transformé. Donc, un meilleur aspect visuel ne garantit pas forcément un meilleur score.


## 10.4 - Random Forest

Cet algorithme d'apprentissage supervisé permet d'entraîner un modèle en utilisant les étiquettes obtenues à partir de la théorie. Il existe un déséquilibre dans les données, car la majorité des étoiles appartiennent aux groupes de la séquence principale et des géantes. Le rapport de classification montre une excellente performance globale du modèle, avec un F1-score proche de 1. Cependant, on observe un déséquilibre notable: les classes majoritaires comme Séquence principale et Géante ont des scores parfaits, tandis que la classe Naine blanche, avec très peu d’exemples (4), présente un rappel faible, ce qui signifie que la moitié des naines blanches n'ont pas été correctement identifiées. Cela reflète l’impact négatif du déséquilibre des classes sur la capacité du modèle à généraliser pour les groupes minoritaires.

## Conclusion et améliorations

Dans ce projet, nous avons comparé plusieurs méthodes de clustering non supervisé (K-Means, DBSCAN, GMM), avec et sans réduction de dimension par PCA, sur des données stellaires représentées par le diagramme HR. Avant d'entraîner nos modèles, nous savions que nous devons avoir 4 clusters : en nous basant sur les diagrammes HR présents dans la littérature scientifique, nous savions que les étoiles se répartissent généralement en quatre grands groupes. Afin de renforcer cette hypothèse et de la valider de manière quantitative, nous avons également appliqué la méthode du coude. Cette dernière a confirmé notre intuition, en indiquant qu’un nombre optimal de clusters égal à 4 était cohérent avec ce projet.

L’algorithme K-Means avec PCA (2 composantes) s’est révélé être le modèle le plus performant, atteignant un silhouette score d’environ 0.35, et offrant une segmentation visuellement cohérente des étoiles. En revanche, DBSCAN s’est montré très sensible à la densité locale et a échoué à détecter une structure exploitable, notamment en test, en assignant la majorité des points au bruit. Le modèle GMM, quant à lui, a connu d’une baisse de performance après PCA, malgré une meilleure visualisation, probablement due à la distorsion des distributions gaussiennes attendues.

Pour compléter cette analyse, nous avons tenté une approche supervisée en générant des labels à partir de nos connaissances sur la physique du diagramme HR. Le modèle de classification obtenu atteint un F1-score global proche de 1, montrant une excellente performance sur les classes majoritaires comme Séquence principale et Géantes. Toutefois, un fort déséquilibre des classes subsiste : des catégories comme les naines blanches, très peu représentées, présentent un rappel très faible, ce qui limite la capacité du modèle à généraliser sur ces groupes rares.

Nous pouvons également proposer des pistes d'amélioration :
